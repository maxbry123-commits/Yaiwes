```python
import json
from datasets import load_dataset
```

## MathQA

```python
math_qa = load_dataset("json", data_files="/mnt/share2/home/zhangxj/workspace/LLM-Adapters/dataset/mathqa/test.json", split="train")
```

```python
math_qa[0]
```

```python
new_math_qa = []
for idx, item in enumerate(math_qa):
    new_item = {
        "id": f"{idx+1:04d}",
        "problem": item["instruction"],
        "answer": item["answer"],
    }
    new_math_qa.append(new_item)

output_file = "/mnt/share2/home/zhangxj/workspace/dynamic_recurrent/evaluation/data/mathqa/test.jsonl"
with open(output_file, 'w', encoding='utf-8') as f:
    for item in new_math_qa:
        json_line = json.dumps(item, ensure_ascii=False)
        f.write(json_line + '\n')

print(f"数据已保存到 {output_file}")
```

## AQuA

```python
aqua = load_dataset("json", data_files="/mnt/share2/home/zhangxj/workspace/LLM-Adapters/dataset/AQuA/test.json", split="train")

aqua[0]
```

```python
new_aqua = []

suffix = "\n\nPlease reason step-by-step and put your choice letter(e.g. A, B, C, D, E) without any other text with \\boxed{} in the end."
for idx, item in enumerate(aqua):
    new_item = {
        "id": f"{idx+1:04d}",
        "problem": item["instruction"] + suffix,
        "solution": item["output"],
        "answer": item["answer"],
    }
    new_aqua.append(new_item)

output_file = "/mnt/share2/home/zhangxj/workspace/dynamic_recurrent/evaluation/data/aqua/test.jsonl"
with open(output_file, 'w', encoding='utf-8') as f:
    for item in new_aqua:
        json_line = json.dumps(item, ensure_ascii=False)
        f.write(json_line + '\n')

print(f"数据已保存到 {output_file}")
```

## ASDiv-Aug

```python
asdiv_aug = load_dataset("json", data_files="/mnt/share2/datasets/ASDiv-Aug/asdiv-aug-test.jsonl", split="train")

asdiv_aug[0]
```

```python
new_asdiv_aug = []
for idx, item in enumerate(asdiv_aug):
    solution = item["answer"]
    answer = solution.split('####')[1].strip()
    new_item = {
        "id": f"{idx+1:04d}",
        "problem": item["question"],
        "solution": item["answer"],
        "answer": answer,
    }
    new_asdiv_aug.append(new_item)

output_file = "/mnt/share2/home/zhangxj/workspace/dynamic_recurrent/evaluation/data/asdiv_aug/test.jsonl"
with open(output_file, 'w', encoding='utf-8') as f:
    for item in new_asdiv_aug:
        json_line = json.dumps(item, ensure_ascii=False)
        f.write(json_line + '\n')

print(f"数据已保存到 {output_file}")
```
